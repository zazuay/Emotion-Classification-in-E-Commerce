```python
import os

# Cari semua file CSV di seluruh input
for root, dirs, files in os.walk("/kaggle/input"):
    for f in files:
        if f.endswith(".csv") and "results" in f:
            print(os.path.join(root, f))
```

```python
import pandas as pd

mbert  = pd.read_csv("/kaggle/input/datasets/zahrapriyono/emotion-classification/Emotion Classification in E-Commerce/results/mbert_results.csv")
xlmr   = pd.read_csv("/kaggle/input/datasets/zahrapriyono/emotion-classification/Emotion Classification in E-Commerce/results/bilstm_results.csv")
bilstm = pd.read_csv("/kaggle/input/datasets/zahrapriyono/emotion-classification/Emotion Classification in E-Commerce/results/xlmr_results.csv")
final = pd.concat([mbert, xlmr, bilstm], ignore_index=True)

# Tambah kolom rank per metric
metrics = ['accuracy', 'weighted_f1', 'macro_f1', 'f1_Happy', 
           'f1_Love', 'f1_Sadness', 'f1_Fear', 'f1_Anger']

for col in metrics:
    final[f'rank_{col}'] = final[col].rank(ascending=False).astype(int)

# Tambah kolom overall rank (rata-rata semua rank)
rank_cols = [f'rank_{col}' for col in metrics]
final['overall_rank'] = final[rank_cols].mean(axis=1).rank().astype(int)

import os

os.makedirs("/kaggle/working/results", exist_ok=True)  # ← tambah ini
final.to_csv("/kaggle/working/results/final_comparison.csv", index=False)
print("Saved: final_comparison.csv")

final.to_csv("/kaggle/working/results/final_comparison.csv", index=False)
print("Saved: final_comparison.csv")
print()

# Tampilkan ringkasan bersih
summary = final[['model'] + metrics + ['overall_rank']].copy()
print(summary.to_string(index=False))
```
